```python
import pandas as pd 
from sklearn.preprocessing import RobustScaler
import pandas as pd



X=pd.read_csv("../data/processed/pima_diabetes_cleaned.csv")
scaler = RobustScaler()

X_scaled = scaler.fit_transform(X)
df_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print("Feature Scaling b RobustScaler activer avec seccées!")
print(df_scaled.head())
df_scaled = df_scaled.round(3)
df_scaled.to_csv("../data/processed/data_Scaled.csv")
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker='o', linestyle='--', color='b')
plt.xlabel('Nombre de clusters (K)')
plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method pour K-Means')
plt.grid(True)
plt.show()
```
